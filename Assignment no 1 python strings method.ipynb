```python

# Python String Methods Assignment

# Step 1: Create 3 variables and take input

name = input("Enter your name: ")
number = input("Enter any number: ")
sentence = input("Enter a sentence: ")


# Step 2: Apply String Methods

print("\n========== STRING METHODS ==========")

# 1. upper()
print("Upper:", name.upper())

# 2. lower()
print("Lower:", name.lower())

# 3. title()
print("Title:", sentence.title())

# 4. capitalize()
print("Capitalize:", sentence.capitalize())

# 5. swapcase()
print("Swapcase:", name.swapcase())

# 6. strip()
print("Strip:", sentence.strip())

# 7. lstrip()
print("Lstrip:", sentence.lstrip())

# 8. rstrip()
print("Rstrip:", sentence.rstrip())

# 9. replace()
print("Replace:", sentence.replace("a", "@"))

# 10. split()
print("Split:", sentence.split())

# 11. join()
print("Join:", "-".join(sentence.split()))

# 12. find()
search_char = name[0] if name else ""
print("Find:", name.find(search_char))

# 13. index()
print("Index:", name.index(search_char))

# 14. count()
print("Count:", sentence.count("a"))

# 15. startswith()
print("Startswith:", sentence.startswith("Hello"))

# 16. endswith()
print("Endswith:", sentence.endswith("."))

# 17. isalpha()
print("Isalpha:", name.isalpha())

# 18. isdigit()
print("Isdigit:", number.isdigit())

# 19. isalnum()
print("Isalnum:", name.isalnum())

# 20. isspace()
print("Isspace:", sentence.isspace())

# 21. islower()
print("Islower:", name.islower())

# 22. isupper()
print("Isupper:", name.isupper())

# 23. istitle()
print("Istitle:", sentence.istitle())

# 24. center()
print("Center:", name.center(20, "*"))

# 25. ljust()
print("Ljust:", name.ljust(20, "*"))

# 26. rjust()
print("Rjust:", name.rjust(20, "*"))

# 27. zfill()
print("Zfill:", number.zfill(10))

# 28. partition()
print("Partition:", sentence.partition(" "))

# 29. rpartition()
print("Rpartition:", sentence.rpartition(" "))

# 30. casefold()
print("Casefold:", name.casefold())


print("\n========== END OF PROGRAM ==========")
```
